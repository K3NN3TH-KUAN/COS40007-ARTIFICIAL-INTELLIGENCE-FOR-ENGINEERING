### Determine Train, Test, Valid Paths

```python
dataset_path = '/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/yolov11'

train_path = os.path.join(dataset_path, 'train')
valid_path = os.path.join(dataset_path, 'valid')
test_path  = os.path.join(dataset_path, 'test')

print(f"Train folder: {train_path}")
print(f"Valid folder: {valid_path}")
print(f"Test folder: {test_path}")
```

### Model Training with yolo11m.pt




```python
from ultralytics import YOLO
import yaml

model = YOLO('yolo11s.pt')

model.train(
    data='/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/yolov11/data.yaml',
    epochs=30,
    imgsz=640,
    batch=16,
    project='car_model_project',
    name='yolo11s_augmented',
    cache=True,
    val=True,
    save=True,
    save_period=1,
    patience=15,
    device=0,
)
```

### Predict Validation Images




```python
results = model.predict(
    source=f"{dataset_path}/test/images",  # Use test images folder
    save=True,
    imgsz=640,
    conf=0.25,
    project="inference_results",
    name="valid_inference"
)
```

### Display Images

```python
from IPython.display import Image, display
from glob import glob

# Corrected path to match `name="valid_inference"`
result_images = glob('/content/inference_results/valid_inference/*.jpg')

# Show 10 example predictions
for img_path in result_images[:10]:
    display(Image(filename=img_path))

print("✅ Displayed some predictions!")
```

### Model Evaluation

```python
val_metrics = model.val(
    data='/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/yolov11/data.yaml',  # Path to data.yaml
    split="val",          # Evaluate on validation set
    imgsz=640,
    conf=0.25
)

print("📊 Validation Results:")
print(val_metrics)

# Run evaluation on the test set
metrics = model.val(
    data='/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/yolov11/data.yaml',
    split='test',
    save_json=True,  # optional: saves COCO-style results for metrics
    save_hybrid=True  # optional: saves hybrid labels
)

print("📊 Testing Results:")
print(metrics)

import shutil
import os

# move test results to your training run folder
test_results_dir = metrics.save_dir  # directory of test results
target_dir = 'car_model_project/yolo11m_augmented/test_results'

if not os.path.exists(target_dir):
    shutil.copytree(test_results_dir, target_dir)

```

### Upload Real-time Images

```python
from google.colab import files
from PIL import Image
from glob import glob
import os

# Upload one or more images
uploaded = files.upload()

# Predict for each uploaded image
for filename in uploaded.keys():
    results = model.predict(
        source=filename,
        save=True,
        imgsz=640,
        conf=0.25
    )

    # Locate the prediction output directory
    save_dir = results[0].save_dir

    # Show first 3 prediction result images
    result_images = glob(os.path.join(save_dir, "*.jpg"))
    for img_path in result_images[:3]:
        display(Image.open(img_path))
```
