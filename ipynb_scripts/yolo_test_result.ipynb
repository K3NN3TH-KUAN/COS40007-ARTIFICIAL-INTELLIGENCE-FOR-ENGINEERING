```python
model_paths = {
    'yolov5': '/content/drive/MyDrive/ModelsEvaluation/Model_Evaluation_Result/model_evaluation_yolov5l/weights/best.pt',
    'yolov8': '/content/drive/MyDrive/ModelsEvaluation/Model_Evaluation_Result/model_evaluation_yolov8l/weights/best.pt',
    'yolov9': '/content/drive/MyDrive/ModelsEvaluation/Model_Evaluation_Result/model_evaluation_yolov9m/weights/best.pt',
    'yolo11': '/content/drive/MyDrive/ModelsEvaluation/Model_Evaluation_Result/model_evaluation_yolo11s/weights/best.pt',
    'yolo12': '/content/drive/MyDrive/ModelsEvaluation/Model_Evaluation_Result/model_evaluation_yolo12s/weights/best.pt',
}

data_yaml = '/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/yolov5/data.yaml'  # must include `test:` path
csv_output_path = '/content/yolo_eval_results.csv'
```

```python
results_list = []

for model_name, model_path in model_paths.items():
    print(f"Evaluating {model_name}...")
    try:
        model = YOLO(model_path)
        metrics = model.val(data=data_yaml, split='test', imgsz=640, iou=0.5)

        result = {
            'Model': model_name,
            'Precision': metrics.results_dict['metrics/precision(B)'],
            'Recall': metrics.results_dict['metrics/recall(B)'],
            'mAP50': metrics.results_dict['metrics/mAP50(B)'],
            'mAP50-95': metrics.results_dict['metrics/mAP50-95(B)'],
            'Fitness': metrics.results_dict['fitness']
        }

    except Exception as e:
        print(f"Error evaluating {model_name}: {e}")
        result = {
            'Model': model_name,
            'Precision': None,
            'Recall': None,
            'mAP50': None,
            'mAP50-95': None,
            'Fitness': None
        }

    results_list.append(result)

# Save to CSV
df = pd.DataFrame(results_list)
df.to_csv(csv_output_path, index=False)
print(f"Saved evaluation results to: {csv_output_path}")
```

```python
import pandas as pd

# Load evaluation results
df = pd.read_csv(csv_output_path)

# Drop rows with missing mAP values (if any model failed)
df_clean = df.dropna(subset=['mAP50-95'])

# Sort by mAP50-95 descending
df_sorted = df_clean.sort_values(by='mAP50-95', ascending=False)

# Get the best model
best_model_row = df_sorted.iloc[0]
best_model_name = best_model_row['Model']
best_map = best_model_row['mAP50-95']

print(f"Best model: {best_model_name} with mAP50-95 = {best_map:.4f}")
```

```python
# Normalize values between 0 and 1 (optional)
df_clean['score'] = (
    0.5 * df_clean['mAP50-95'] +
    0.3 * df_clean['Precision'] +
    0.2 * df_clean['Recall']
)

best_model_row = df_clean.sort_values(by='score', ascending=False).iloc[0]
print(f"Best model (custom score): {best_model_row['Model']}")
```
