```python
## Import necessary libraries
from ultralytics import YOLO
import cv2
import numpy as np
import tempfile
import os
import gradio as gr

# Model paths
model_paths = {
    "YOLOv5": "/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/best_model/best_yolov5lu.pt",
    "YOLOv8": "/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/best_model/best_yolov8l.pt",
    "YOLOv9": "/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/best_model/best_yolov9m.pt",
    "YOLOv11": "/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/best_model/best_yolo11s.pt",
    "YOLOv12": "/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/best_model/best_yolo12s.pt"
}

DEFAULT_CONF_THRESHOLD = 0.50 # Default confidence threshold for detections

CLASS_COLORS = {
    0: (255, 0, 0),
    1: (0, 255, 0),
    2: (0, 0, 255),
    3: (255, 255, 0),
    4: (255, 0, 255),
}

current_model = None
current_model_name = None

# Load the YOLO model based on the selected model name
def load_model(model_name):
    global current_model, current_model_name
    if model_name not in model_paths:
        raise ValueError("Please select a valid YOLO model.")
    if model_name != current_model_name:
        current_model = YOLO(model_paths[model_name])
        current_model_name = model_name

# Detect anomalies based on car count and detection boxes
def detect_anomaly(car_count, all_boxes, conf_threshold=DEFAULT_CONF_THRESHOLD):
    allowed_classes = {0, 1, 2, 3, 4}
    unexpected_class_counts = {}
    low_conf_count = 0

    for box in all_boxes:
        cls_id = int(box.cls[0])
        conf = float(box.conf[0])
        if cls_id not in allowed_classes:
            unexpected_class_counts[cls_id] = unexpected_class_counts.get(cls_id, 0) + 1
        if conf < conf_threshold:
            low_conf_count += 1

    messages = []

    if car_count > 10:
        messages.append(f"🚨 Too Many Cars detected: {car_count} cars (limit exceeded).")
    elif car_count == 0:
        messages.append("ℹ️ No cars detected.")
    else:
        messages.append("✅ Car count is normal.")

    if low_conf_count > 0:
        messages.append(f"⚠️ {low_conf_count} low-confidence detection(s) found.")

    if unexpected_class_counts:
        unexpected_msgs = []
        for cls_id, count in unexpected_class_counts.items():
            name = current_model.names[cls_id] if cls_id < len(current_model.names) else f"Class {cls_id}"
            unexpected_msgs.append(f"{name} (class {cls_id}): {count}")
        messages.append("🚫 Unexpected vehicle classes detected:\n" + "\n".join(unexpected_msgs))

    return "\n".join(messages)

# Draw bounding boxes on the image
# Annotate the image with bounding boxes, labels, and confidence scores
def draw_boxes(image, result, conf_threshold=DEFAULT_CONF_THRESHOLD):
    annotated = image.copy()
    LOW_CONF_COLOR = (128, 128, 128)

    for box in result.boxes:
        conf = float(box.conf[0])
        cls_id = int(box.cls[0])
        xyxy = box.xyxy[0].cpu().numpy().astype(int)
        label = current_model.names[cls_id]

        color = LOW_CONF_COLOR if conf < conf_threshold else CLASS_COLORS.get(cls_id, (255, 255, 255))

        cv2.rectangle(annotated, (xyxy[0], xyxy[1]), (xyxy[2], xyxy[3]), color, 4)
        label_text = f"{label} ({conf:.2f})"
        font = cv2.FONT_HERSHEY_SIMPLEX
        font_scale = 1.5
        font_thickness = 2
        (text_width, text_height), _ = cv2.getTextSize(label_text, font, font_scale, font_thickness)
        text_bg_topleft = (xyxy[0], max(xyxy[1] - text_height - 10, 0))
        text_bg_bottomright = (xyxy[0] + text_width, xyxy[1])
        cv2.rectangle(annotated, text_bg_topleft, text_bg_bottomright, (255, 255, 255), -1)
        cv2.putText(annotated, label_text, (xyxy[0], xyxy[1] - 5), font, font_scale, color, font_thickness, cv2.LINE_AA)

    return annotated

# Detect on Image
def detect_on_image(image, model_choice, conf_threshold=DEFAULT_CONF_THRESHOLD):
    if model_choice is None:
        return np.zeros((512, 512, 3), dtype=np.uint8), "❌ Please select a YOLO model.", "Model not selected."

    try:
        load_model(model_choice)
        results = current_model(image)
        result = results[0]

        all_boxes = result.boxes
        car_classes = {0, 1, 2, 3, 4}
        class_counts = {cls: 0 for cls in car_classes}
        car_count = 0

        for box in all_boxes:
            cls_id = int(box.cls[0])
            if cls_id in car_classes:
                class_counts[cls_id] += 1
                car_count += 1

        car_count_summary = f"Total: {car_count} cars\n" + "\n".join(
            [f"Class {cls_id} ({current_model.names[cls_id]}): {count}" for cls_id, count in class_counts.items()]
        )

        annotated = draw_boxes(image, result, conf_threshold)
        anomaly = detect_anomaly(car_count, all_boxes, conf_threshold)

        return annotated, car_count_summary, anomaly

    except Exception as e:
        print("Image error:", e)
        return np.zeros((512, 512, 3), dtype=np.uint8), "Error", str(e)

# Detect on Video
def detect_on_video(video_file, model_choice, conf_threshold=DEFAULT_CONF_THRESHOLD):
    if model_choice is None:
        return None, "❌ Please select a YOLO model.", "Model not selected."

    try:
        load_model(model_choice)
        input_path = video_file
        temp_dir = tempfile.mkdtemp()
        output_path = os.path.join(temp_dir, "output.mp4")

        cap = cv2.VideoCapture(input_path)
        width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
        fps = cap.get(cv2.CAP_PROP_FPS)

        fourcc = cv2.VideoWriter_fourcc(*'mp4v')
        out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

        total_car_count = 0
        frame_count = 0
        all_boxes_accumulated = []

        car_classes = {0, 1, 2, 3, 4}
        class_counts_total = {cls: 0 for cls in car_classes}

        while cap.isOpened():
            ret, frame = cap.read()
            if not ret:
                break
            frame_count += 1
            results = current_model(frame)
            result = results[0]
            all_boxes = result.boxes

            class_counts = {cls: 0 for cls in car_classes}
            car_count = 0

            for box in all_boxes:
                cls_id = int(box.cls[0])
                if cls_id in car_classes:
                    class_counts[cls_id] += 1
                    car_count += 1
                all_boxes_accumulated.append(box)

            for cls_id in class_counts:
                class_counts_total[cls_id] += class_counts[cls_id]

            total_car_count += car_count
            annotated = draw_boxes(frame, result, conf_threshold)
            out.write(annotated)

        cap.release()
        out.release()

        class_summary_lines = [
            f"Class {cls} ({current_model.names[cls]}): Total {class_counts_total[cls]}, Avg/frame {class_counts_total[cls]/frame_count:.2f}"
            for cls in class_counts_total
        ]
        car_count_summary = f"Total Cars Detected: {total_car_count} over {frame_count} frames\n" + "\n".join(class_summary_lines)
        anomaly = detect_anomaly(total_car_count, all_boxes_accumulated, conf_threshold)

        return output_path, car_count_summary, anomaly

    except Exception as e:
        print("Video error:", e)
        return None, "Error", str(e)

# Gradio Image Interface
image_interface = gr.Interface(
    fn=detect_on_image,
    inputs=[
        gr.Image(type="numpy", label="Upload Image", height=480, width=720),
        gr.Dropdown(choices=list(model_paths.keys()), label="Select YOLO Model", value=None, interactive=True),
        gr.Slider(minimum=0.1, maximum=1.0, step=0.05, value=DEFAULT_CONF_THRESHOLD, label="Confidence Threshold")
    ],
    outputs=[
        gr.Image(label="Detected Image", height=480, width=720),
        gr.Textbox(label="🚘 Car Detection Summary"),
        gr.Textbox(label="⚠️ Anomaly Detection")
    ],
    title="🚗 Car Model Detection (Image)"
)

# Gradio Video Interface
video_interface = gr.Interface(
    fn=detect_on_video,
    inputs=[
        gr.Video(label="Upload Video", height=480, width=720),
        gr.Dropdown(choices=list(model_paths.keys()), label="Select YOLO Model", value=None, interactive=True),
        gr.Slider(minimum=0.1, maximum=1.0, step=0.05, value=DEFAULT_CONF_THRESHOLD, label="Confidence Threshold")
    ],
    outputs=[
        gr.Video(label="Detected Video", height=480, width=720),
        gr.Textbox(label="🚘 Car Detection Summary"),
        gr.Textbox(label="⚠️ Anomaly Detection")
    ],
    title="🚗 Car Model Detection (Video)"
)

# Launch Gradio with Tabs
gr.TabbedInterface(
    [image_interface, video_interface],
    ["Image Detection", "Video Detection"]
).launch(share=True)
```
