```python
dataset_path = '/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/yolov8'

train_path = os.path.join(dataset_path, 'train')
valid_path = os.path.join(dataset_path, 'valid')
test_path  = os.path.join(dataset_path, 'test')

print(f"Train folder: {train_path}")
print(f"Valid folder: {valid_path}")
print(f"Test folder: {test_path}")
```

```python
from ultralytics import YOLO
import yaml

# Create model - Use a YOLOv5 pre-trained weight file
model = YOLO('yolov8l.pt')

model.train(
    data='/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/yolov8/data.yaml',
    epochs=30,
    imgsz=640,
    batch=16,
    project='car_model_project2',
    name='yolov8l_augmented',
    cache=True,
    val=True,
    save=True,
    save_period=1,
    patience=15,
    device=0,
)
```

```python
# ===============================
# Step 6: Run Inference on valid Set
# ===============================
results = model.predict(
    source=f"{dataset_path}/valid/images",  # Use validation images folder
    save=True,
    imgsz=640,
    conf=0.25,
    project="inference_results",
    name="valid_inference"
)
```

```python
# ===========================
# Step 8: Show Inference Images
# ===========================
from IPython.display import Image, display
from glob import glob

# Corrected path to match `name="valid_inference"`
result_images = glob('/content/inference_results/valid_inference/*.jpg')

# Show 5 example predictions
for img_path in result_images[:10]:
    display(Image(filename=img_path))

print("✅ Displayed some predictions!")
```

```python
# Evaluate the model using validation set with labels
val_metrics = model.val(
    data='/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/yolov5/data.yaml',  # Path to data.yaml
    split="val",          # Evaluate on validation set
    imgsz=640,
    conf=0.25
)

print("📊 Validation Results:")
print(val_metrics)
```

```python
# ===============================
# Step 6: Run Inference on test Set
# ===============================
results = model.predict(
    source=f"{dataset_path}/test/images",  # Use test images folder
    save=True,
    imgsz=640,
    conf=0.25,
    project="testinference_results",
    name="test_inference"
)
```

```python
# ===========================
# Step 8: Show Inference Images
# ===========================
from IPython.display import Image, display
from glob import glob

# Corrected path to match `name="test_inference"`
result_images = glob('/content/inference_results/test_inference/*.jpg')

# Show 5 example predictions
for img_path in result_images[:10]:
    display(Image(filename=img_path))

print("✅ Displayed some predictions!")
```

```python
test_metrics = model.val(
    data='/content/COS40007-ARTIFICIAL-INTELLIGENCE-FOR-ENGINEERING/yolov5/data.yaml',
    split='test',
    imgsz=640,
    conf=0.25,
    save=True,
    save_json=True
)

print("🧪 Test Results:")
print(test_metrics)
```

```python
# ============================
# Step 9: Real-World Upload and Live Prediction
# ============================

from google.colab import files
from PIL import Image
from glob import glob
import os

# Upload one or more images
uploaded = files.upload()

# Predict for each uploaded image
for filename in uploaded.keys():
    results = model.predict(
        source=filename,
        save=True,
        imgsz=640,
        conf=0.25
    )

    # Locate the prediction output directory
    save_dir = results[0].save_dir

    # Show first 3 prediction result images
    result_images = glob(os.path.join(save_dir, "*.jpg"))
    for img_path in result_images[:3]:
        display(Image.open(img_path))
```
